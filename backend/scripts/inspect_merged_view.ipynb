```python
import psycopg2
import pandas as pd
from dotenv import load_dotenv
import os
load_dotenv()  # load environment variables from .env file
```

```python
DB_HOST="localhost"
DB_NAME="renewable_db"
DB_USER=os.getenv('USERNAME')
DB_PASSWORD=os.getenv('PASSWORD')
DB_PORT=5432
VIEW_NAME="merged_weather_solar_view"
conn=psycopg2.connect(
    host=DB_HOST,
    database=DB_NAME,
    user=DB_USER,
    password=DB_PASSWORD,
    port=DB_PORT
)
try:
    # select all data from the merged view to verify it was created successfully
    query=f"SELECT * FROM {VIEW_NAME};"
    df=pd.read_sql(query, conn)
    print(f"Successfully retrieved {len(df)} rows from {VIEW_NAME}")
finally:
    conn.close()
df.to_csv("sample_merged_view_data.csv", index=False)
df.head(10)
```
